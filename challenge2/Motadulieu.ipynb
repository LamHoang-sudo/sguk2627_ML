# BÀI TẬP PHÂN TÍCH KHÁM PHÁ DỮ LIỆU RƯỢU ĐỎ (WINE QUALITY)

### 1. Import thư viện và nạp dữ liệu vào notebook

```python
import numpy as np
import pandas as pd
from scipy import stats

# Nạp tập dữ liệu rượu đỏ
wine_data = pd.read_csv('/content/winequality-red.csv')

# Kiểm tra sơ bộ dữ liệu
print(f"Kích thước tập dữ liệu ban đầu: {wine_data.shape}")
display(wine_data.head())
```

### 2. Loại bỏ dữ liệu trùng lặp

```python
# Kiểm tra số lượng dòng trùng lặp
duplicates_count = wine_data.duplicated().sum()
print(f"Số dòng trùng lặp phát hiện: {duplicates_count}")

# Loại bỏ dòng trùng lặp
wine_data = wine_data.drop_duplicates()
print(f"Kích thước sau khi loại bỏ trùng lặp: {wine_data.shape}")
```

### 3. Thay thế dữ liệu và thay đổi định dạng của dữ liệu

```python
# Chuyển đổi cột 'quality' về kiểu dữ liệu số nguyên chuẩn
wine_data['quality'] = wine_data['quality'].astype(int)

# Chuyển đổi các cột số thực sang kiểu float32 để tiết kiệm bộ nhớ
float_cols = wine_data.select_dtypes(include=['float64']).columns
wine_data[float_cols] = wine_data[float_cols].astype('float32')

# Hiển thị cấu trúc kiểu dữ liệu mới
display(wine_data.dtypes)
```

### 4. Xử lý dữ liệu thiếu và tính toán các đặc trưng thống kê mô tả

```python
# Kiểm tra và xử lý dữ liệu thiếu (điền khuyết bằng trung vị của mỗi cột nếu có giá trị Null)
for col in wine_data.columns:
    missing_count = wine_data[col].isnull().sum()
    if missing_count > 0:
        median_val = wine_data[col].median()
        wine_data[col].fillna(median_val, inplace=True)
        print(f"Đã xử lý cột '{col}': điền {missing_count} giá trị thiếu bằng trung vị {median_val}")

# Lấy cột dữ liệu 'quality' để phân tích thống kê
target_col = wine_data['quality'].values

# Tính toán các thông số thống kê bằng numpy và scipy.stats
data_mean = np.mean(target_col)
data_median = np.median(target_col)
data_mode = stats.mode(target_col, keepdims=True)
data_variance = np.var(target_col)
data_sd = np.std(target_col)
data_min = np.min(target_col)
data_max = np.max(target_col)
data_range = data_max - data_min
data_percentile_60 = np.percentile(target_col, 60)
data_q1 = np.percentile(target_col, 25)
data_q3 = np.percentile(target_col, 75)
data_IQR = stats.iqr(target_col)

# In kết quả kiểm tra
print("=== KẾT QUẢ THỐNG KÊ MÔ TẢ TRÊN CỘT 'QUALITY' ===")
print(f"Mean (Trung bình):                  {data_mean:.4f}")
print(f"Median (Trung vị):                  {data_median}")
print(f"Mode (Yếu vị):                      {data_mode.mode[0]} (Xuất hiện {data_mode.count[0]} lần)")
print(f"Variance (Phương sai):              {data_variance:.4f}")
print(f"Standard Deviation (Độ lệch chuẩn): {data_sd:.4f}")
print(f"Range (Khoảng biến thiên):          {data_range} (Min: {data_min}, Max: {data_max})")
print(f"60th Percentile (Phân vị 60):       {data_percentile_60}")
print(f"Quartiles:                          Q1 (25%): {data_q1}, Q3 (75%): {data_q3}")
print(f"IQR (Khoảng tứ phân vị):            {data_IQR}")
print("===================================================")
```
